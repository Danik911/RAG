````python
import re
import docx

def remove_markdown(text):
    """
    Remove common Markdown syntax from text.
    """
    # Remove Markdown headers (e.g., "# Header")
    text = re.sub(r'^\s*#+\s*', '', text, flags=re.MULTILINE)

    # Replace links [text](url) with just 'text'
    text = re.sub(r'\[([^\]]+)\]\(([^)]+)\)', r'\1', text)

    # Replace images ![alt](url) with the alt text only
    text = re.sub(r'!\[([^\]]*)\]\(([^)]+)\)', r'\1', text)

    # Remove inline code markers (backticks)
    text = re.sub(r'`([^`]+)`', r'\1', text)

    # Remove code blocks delimited by triple backticks
    text = re.sub(r'```.*?```', '', text, flags=re.DOTALL)

    # Remove emphasis markers: *, **, _, __
    text = re.sub(r'(\*\*|__|\*|_)', '', text)

    # Remove blockquote markers (leading '>')
    text = re.sub(r'^\s*>\s?', '', text, flags=re.MULTILINE)

    # Remove list markers (e.g., "-", "*", "+")
    text = re.sub(r'^\s*[-*+]\s+', '', text, flags=re.MULTILINE)

    return text

def read_docx(file_path):
    """
    Read all paragraphs from a DOCX file and return a single string.
    """
    doc = docx.Document(file_path)
    full_text = []
    for para in doc.paragraphs:
        full_text.append(para.text)
    return '\n'.join(full_text)

def main():
    # Define your file path (note the raw string literal to handle backslashes)
    file_path = r"C:\Users\anteb\OneDrive\Desktop\Griffith\Business\assignment\Kotter's 8.docx"

    # Read text from the DOCX file
    text = read_docx(file_path)

    # Remove Markdown formatting
    cleaned_text = remove_markdown(text)

    # Print the cleaned text to the notebook output
    print(cleaned_text)

    # Optionally, write the cleaned text to a new file
    with open("cleaned_text.txt", "w", encoding="utf-8") as f:
        f.write(cleaned_text)

if __name__ == "__main__":
    main()
````
